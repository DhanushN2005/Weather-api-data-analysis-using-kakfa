```python
import psycopg2
import pandas as pd

conn = psycopg2.connect(
    dbname="weatherdb",
    user="root",
    password="root",
    host="127.0.0.1",
    port="5433"
)

df = pd.read_sql("SELECT city, temperature, humidity FROM weather_data", conn)
conn.close()

def classify(temp):
    if temp >= 35:
        return "VERY HOT"
    elif temp >= 30:
        return "HOT"
    elif temp >= 25:
        return "WARM"
    else:
        return "COOL"

df["weather_class"] = df["temperature"].apply(classify)

print(df[["city", "temperature", "humidity", "weather_class"]].head())
```

```python
import psycopg2
import pandas as pd
from sklearn.cluster import KMeans

# 🔗 Connect to PostgreSQL (Docker)
conn = psycopg2.connect(
    dbname="weatherdb",
    user="root",
    password="root",
    host="127.0.0.1",
    port="5433"   # use 5432 if your postgres is on 5432
)

# 📥 Load data from DB
df = pd.read_sql(
    "SELECT city, temperature, humidity FROM weather_data",
    conn
)
conn.close()

# 🧠 Features for ML
X = df[["temperature", "humidity"]]

# 🤖 KMeans clustering
# With low variance data, 1 cluster is mathematically correct
kmeans = KMeans(n_clusters=1, random_state=42)
df["cluster"] = kmeans.fit_predict(X)

# ✅ Output
print(df.head())
print("\nCluster centers:")
print(kmeans.cluster_centers_)
```
